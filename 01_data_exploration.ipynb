```python
import pandas as pd

# 1. Target Distribution (Class Imbalance)
print("--- ESI Level Distribution ---")
esi_counts = df['esi_level'].value_counts().sort_index()
print(pd.DataFrame({'Count': esi_counts, 'Percentage (%)': (esi_counts / len(df) * 100).round(2)}))

# 2. Missing Values Analysis
print("\n--- Missing Values ---")
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing (%)': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

# 3. Text Column Coverage
print("\n--- Text Feature Availability ---")
print("Missing Chief Complaints:", df['chief_complaint'].isnull().sum())
print("Missing Clinical Notes:", df['clinical_notes'].isnull().sum())
```
