```python
import pandas as pd

df = pd.read_csv("fedmml_ed_triage_dataset.csv")

df.info()
```

```python
print(df.columns.tolist())
```

```python
print("--- ESI Class Distribution ---")
esi_dist = df['esi_level'].value_counts(normalize=True).sort_index() * 100
for level, pct in esi_dist.items():
    print(f"ESI Level {level}: {pct:.2f}%")

# 3. Missing Values Breakdown
print("\n--- Missing Values Summary ---")
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing, 'Percentage': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

# 4. Text Length Analysis for Bio-ClinicalBERT Tokenization
text_cols = ['chief_complaint', 'clinical_notes']
for col in text_cols:
    df[f'{col}_word_count'] = df[col].fillna('').astype(str).apply(lambda x: len(x.split()))
    print(f"\n--- {col} Word Count Stats ---")
    print(df[f'{col}_word_count'].describe(percentiles=[0.5, 0.9, 0.95, 0.99]))
```
